字符串的乐趣

```python
print("\n第一部分")
def is_palindrome(text):
    cleaned = text.replace(" ", "").lower()
    return cleaned == cleaned[::-1]

test_words = ["level", "tacocat", "T A C O cat", "rats live on no evil star", "hello", "abc"]
for s in test_words:
    print(f"{s!r:30s} -> {is_palindrome(s)}")

print("\n第二部分")
def is_abecedarian(text):
    letters = text.replace(" ", "").lower()
    for i in range(len(letters) - 1):
        if letters[i] > letters[i + 1]:
            return False
    return True

test_words = ["adder", "beet", "abcd efgh xyz", "almost", "dog", "cat", "zoo"]
for s in test_words:
    print(f"{s!r:16s} -> {is_abecedarian(s)}")

print("\n第三部分")
def remove_vowels(text):
    vowels = "aeiouAEIOU"
    return "".join(ch for ch in text if ch not in vowels)

test_words = ["cat", "audio", "aeiou", "rhythm", "Hello World", "PYTHON"]
for s in test_words:
    print(f"{s!r:14s} -> {remove_vowels(s)!r}")
```

列表的乐趣

```python
print("\n第一部分")
def list_reverse(seq):
    if not isinstance(seq, list):
        raise TypeError(f"期望 list，实际收到 {type(seq).__name__}")
    return seq[::-1]


for value in ([1, 2, 3], ["a", "b"], [], [1], (1, 2, 3), "abc", 5):
    try:
        print(f"{value!r:16s} -> {list_reverse(value)!r}")
    except TypeError as e:
        print(f"{value!r:16s} -> TypeError: {e}")

print("\n第二部分")
def is_sorted(seq):
    for i in range(len(seq) - 1):
        if seq[i] > seq[i + 1]:
            return False
    return True

for value in ([1, 2, 3, 4], [1, 1, 2], [], [7], [1, 3, 2], ["a", "b", "c"], "abcd", (1, 2, 3)):
    print(f"{value!r:18s} -> {is_sorted(value)}")
```

作为向量的元组

```python
NUMBER_TYPES = (int, float)

print("\n第一部分")
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError(f"向量必须是元组，实际收到 {type(t).__name__}")
    if not isinstance(s, NUMBER_TYPES):
        raise TypeError(f"标量必须是整数或浮点数，实际收到 {type(s).__name__}")
    for j, v in enumerate(t):
        if not isinstance(v, NUMBER_TYPES):
            raise TypeError(f"元组第 {j} 个元素不是数值，实际收到 {type(v).__name__}")
    return tuple(v * s for v in t)

for t, s in [((1, 2, 3), 2), ((1.5, -2), 0.5), ((), 3), ((1, 2), 2.5)]:
    print(f"vec_scalar_mult({t}, {s}) = {vec_scalar_mult(t, s)}")

for bad in [([1, 2], 2), ((1, 2), "2"), ((1, "a"), 2)]:
    try:
        vec_scalar_mult(*bad)
    except TypeError as e:
        print(f"  {bad} -> TypeError: {e}")

print("\n第二部分")
def vec_inner_product(x, y):
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError(f"两个参数都必须为元组，实际收到 {type(x).__name__} 和 {type(y).__name__}")
    if len(x) != len(y):
        raise ValueError(f"向量长度必须一致，实际为 {len(x)} 和 {len(y)}")
    for j, (a, b) in enumerate(zip(x, y)):
        if not isinstance(a, NUMBER_TYPES) or not isinstance(b, NUMBER_TYPES):
            raise TypeError(f"第 {j} 个分量不是数值")
    return float(sum(a * b for a, b in zip(x, y)))

for x, y in [((1, 2, 3), (4, 5, 6)), ((1.5, 2), (2, 4)), ((1,), (7,)), ((), ())]:
    print(f"vec_inner_product({x}, {y}) = {vec_inner_product(x, y)}")

print()
for args in [([1, 2], (1, 2)), ((1, 2), [1, 2]), ((1, 2), (1, 2, 3))]:
    try:
        vec_inner_product(*args)
    except (TypeError, ValueError) as e:
        print(f"vec_inner_product({args[0]}, {args[1]}) -> {type(e).__name__}: {e}")

print("\n第三部分")
def check_valid_mx(m):
    if not isinstance(m, tuple):
        return False
    for row in m:
        if not isinstance(row, tuple):
            return False
    lengths = {len(row) for row in m}
    if len(lengths) > 1:
        return False
    for row in m:
        for v in row:
            if not isinstance(v, NUMBER_TYPES):
                return False
    return True

cases = [((1, 2), (3, 4)), ((1, 2), (3, 4, 5)), ((1, 2), [3, 4]),
         [[1, 2], (3, 4)], ((1, "a"), (3, 4)), ((1, 2), (3, "b")),
         ((), ()), (), ((1,),), ((1.5, 2), (3, 4))]
for c_ in cases:
    print(f"  {c_!r:26s} -> {check_valid_mx(c_)}")

print("\n第四部分")
def mx_vec_mult(m, v):
    if not check_valid_mx(m):
        raise TypeError("第一个参数不是合法的矩阵（应为元组的元组，行等长，元素均为数值）")
    if not isinstance(v, tuple):
        raise TypeError(f"向量必须是元组，实际收到 {type(v).__name__}")
    ncols = len(m[0]) if m else 0
    if len(v) != ncols:
        raise ValueError(f"向量长度 {len(v)} 与矩阵列数 {ncols} 不一致")
    for j, val in enumerate(v):
        if not isinstance(val, NUMBER_TYPES):
            raise TypeError(f"向量第 {j} 个分量不是数值")
    return tuple(sum(row[j] * v[j] for j in range(ncols)) for row in m)

M = ((1, 2), (3, 4))
v = (5, 6)

print(f"矩阵 M = {M}")
print(f"向量 v = {v}")
print(f"mx_vec_mult(M, v) = {mx_vec_mult(M, v)}")
print("手算验证: (1*5+2*6, 3*5+4*6) = (17, 39)")

print()
for args in [(v, v), (M, [5, 6]), (M, (5, 6, 7)), (((1, 2), (3,)), (1, 2))]:
    try:
        print(f"mx_vec_mult({args[0]}, {args[1]}) = {mx_vec_mult(*args)}")
    except (TypeError, ValueError) as e:
        print(f"mx_vec_mult({args[0]}, {args[1]}) -> {type(e).__name__}: {e}")
```

更多向量的乐趣

```python

NUMBER_TYPES = (int, float)

def check_number(value, name):
    if isinstance(value, bool) or not isinstance(value, NUMBER_TYPES):
        raise TypeError(f"{name}必须是整数或浮点数，实际收到 {type(value).__name__}")


class Vector:
    def __init__(self, dim, elements=None):
        if isinstance(dim, bool) or not isinstance(dim, int):
            raise TypeError("维度必须是非负整数")
        if dim < 0:
            raise ValueError("维度不能为负数")

        if elements is None:
            elements = [0] * dim

        if not isinstance(elements, (list, tuple)):
            raise TypeError("向量元素必须用列表或元组提供")
        if len(elements) != dim:
            raise ValueError(f"维度是 {dim}，但提供了 {len(elements)} 个元素")

        for i, value in enumerate(elements):
            check_number(value, f"第 {i} 个元素")

        self._dim = dim
        self._elements = tuple(elements)

    @property
    def dim(self):
        return self._dim

    @property
    def elements(self):
        return self._elements

    def __repr__(self):
        return f"Vector({self._dim}, {self._elements})"

    def _check_comparable(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能与另一个 Vector 对象比较")
        if self._dim != other._dim:
            raise ValueError(f"向量维度不一致：{self._dim} 和 {other._dim}")

    # 第三部分：比较运算符，按元素元组的字典序比较
    def __eq__(self, other):
        self._check_comparable(other)
        return self._elements == other._elements

    def __ne__(self, other):
        self._check_comparable(other)
        return self._elements != other._elements

    def __lt__(self, other):
        self._check_comparable(other)
        return self._elements < other._elements

    def __le__(self, other):
        self._check_comparable(other)
        return self._elements <= other._elements

    def __gt__(self, other):
        self._check_comparable(other)
        return self._elements > other._elements

    def __ge__(self, other):
        self._check_comparable(other)
        return self._elements >= other._elements

    # 第四部分：向量内积
    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("dot() 的参数必须是 Vector 对象")
        if self._dim != other._dim:
            raise ValueError(f"向量维度不一致：{self._dim} 和 {other._dim}")
        return sum(a * b for a, b in zip(self._elements, other._elements))

    # 第五部分：标量乘法或向量逐元素乘法
    def __mul__(self, other):
        if isinstance(other, Vector):
            if self._dim != other._dim:
                raise ValueError(f"向量维度不一致：{self._dim} 和 {other._dim}")
            result = [a * b for a, b in zip(self._elements, other._elements)]
            return Vector(self._dim, result)

        check_number(other, "乘数")
        result = [value * other for value in self._elements]
        return Vector(self._dim, result)

    def __rmul__(self, other):
        return self * other


# 第一部分测试：创建向量和检查无效输入
print("\n第一部分：创建向量")
print(Vector(3, (1, 2, 3)))
print(Vector(3))
print(Vector(2, [1.5, -2]))

for args in [(-1,), ("3",), (3, (1, 2)), (2, 5), (2, (1, "a"))]:
    try:
        Vector(*args)
    except (TypeError, ValueError) as e:
        print(f"Vector{args} -> {type(e).__name__}: {e}")


# 第二部分说明和测试：维度、元素属于每个对象
print("\n第二部分：实例属性")
v1 = Vector(2, (1, 2))
v2 = Vector(3, (7, 8, 9))
print("v1:", v1, "| dim:", v1.dim, "| elements:", v1.elements)
print("v2:", v2, "| dim:", v2.dim, "| elements:", v2.elements)


# 第三部分测试：比较运算符
print("\n第三部分：比较运算符")
pairs = [
    (Vector(2, (2, 4)), Vector(2, (2, 5))),
    (Vector(2, (2, 4)), Vector(2, (2, 4))),
    (Vector(3, (1, 2, 3)), Vector(3, (1, 2, 4))),
]

for a, b in pairs:
    print(f"a = {a}, b = {b}")
    print(f"  a==b: {a == b}")
    print(f"  a!=b: {a != b}")
    print(f"  a<b:  {a < b}")
    print(f"  a<=b: {a <= b}")
    print(f"  a>b:  {a > b}")
    print(f"  a>=b: {a >= b}")

try:
    print(Vector(2, (1, 2)) < Vector(3, (1, 2, 3)))
except ValueError as e:
    print("不同维度比较 ->", e)


# 第四部分测试：dot 内积
print("\n第四部分：dot 内积")
a = Vector(3, (1, 2, 3))
b = Vector(3, (4, 5, 6))
print(f"{a}.dot({b}) = {a.dot(b)}")

try:
    a.dot(Vector(2, (1, 2)))
except ValueError as e:
    print("不同维度内积 ->", e)


# 第五部分测试：标量乘法和向量逐元素乘法
print("\n第五部分：乘法")
v = Vector(2, (2, 3))
w = Vector(2, (4, 5))

print(f"2 * {v} = {2 * v}")
print(f"{v} * 2 = {v * 2}")
print(f"{v} * {w} = {v * w}")

try:
    v * Vector(3, (1, 2, 3))
except ValueError as e:
    print("不同维度向量相乘 ->", e)
```

第二部分\
维度和元素应设为实例属性，理由：
1. 每个向量对象都有自己独立的维度和元素。若设为类属性（定义在 class 体内、__init__ 之外），
   所有 Vector 实例会共享同一份数据，修改一个对象会影响其他所有对象；
2. 不同向量可以有不同维度（如 Vector(2, ...) 与 Vector(3, ...)），类属性无法表达这种差异；
3. 实例属性在 __init__ 中用 self._dim / self._elements 赋值，每个对象在自己的
   __dict__ 中各自保存一份，互不干扰；
4. 对外通过 @property 只读访问，避免外部随意修改导致维度与元素个数不一致。
