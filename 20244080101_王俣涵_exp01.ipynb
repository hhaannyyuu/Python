实验2：函数、条件、循环与递归

```python
#实验二
print("\n第一部分")
def score_level(score):
    if not 0 <= score <= 100:
        raise ValueError(f"分数必须在 0~100 之间，收到: {score}")
    if score >= 90:
        return "优秀"
    if score >= 80:
        return "良好"
    if score >= 60:
        return "及格"
    return "不及格"

print("【正常分数】")
for s in (100, 95, 90, 85, 80, 75, 60, 59, 0):
    print(f"{s:>3} 分 -> {score_level(s)}")

print("【越界分数】")
for s in (-1, 101):
    try:
        score_level(s)
    except ValueError as e:
        print(f"捕获到异常: {e}")

print("\n第二部分")
total_for = 0
for i in range(1, 101):
    total_for += i

total_while = 0
i = 1
while i <= 100:
    total_while += i
    i += 1

print(f"for 循环结果:   {total_for}")
print(f"while 循环结果: {total_while}")
print(f"两种实现结果相同: {total_for == total_while}")

print("\n第三部分")
def factorial_recursive(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("n 必须是非负整数")
    if n <= 1:
        return 1
    return n * factorial_recursive(n - 1)

def factorial_iterative(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("n 必须是非负整数")
    result = 1
    for k in range(2, n + 1):
        result *= k
    return result

print(f"{'n':>3} | {'递归':>10} | {'迭代':>10} | 结果一致")
print("-" * 42)
for n in (0, 1, 5, 10):
    r, it = factorial_recursive(n), factorial_iterative(n)
    print(f"{n:>3} | {r:>10} | {it:>10} | {r == it}")

print("\n第四部分")
def summary(values):
    if not values:
        raise ValueError("values 不能为空列表")
    return (min(values), max(values), sum(values) / len(values))

data = [3, 1, 4, 1, 5, 9, 2, 6]
lowest, highest, average = summary(data)  # 元组解包
print(f"列表: {data}")
print(f"最小值 = {lowest}, 最大值 = {highest}, 平均值 = {average:.4f}")

print("\n第五部分")
def sum_all(*args):
    return sum(args)

print(f"无参数:   sum_all() = {sum_all()}")
print(f"单个参数: sum_all(5) = {sum_all(5)}")
print(f"多个参数: sum_all(1, 2, 3, 4, 5) = {sum_all(1, 2, 3, 4, 5)}")
```

实验3：列表、字典、元组与基础统计

```python
#实验三
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]
print("第一部分")
name_to_score = dict(zip(names, scores))
print(f"姓名->成绩字典: {name_to_score}")
for name, score in name_to_score.items():
    print(f"{name}: {score}")

print("\n第二部分")
for index, (name, score) in enumerate(name_to_score.items(), start=1):
    print(f"第 {index} 位: {name} -> {score} 分")

print("\n第三部分")
passed_names = [name for name, score in zip(names, scores) if score >= 60]
squared_scores = [score ** 2 for score in scores]
print(f"及格学生: {passed_names}")
print(f"成绩平方: {squared_scores}")

print("\n第四部分")
count = len(scores)
average = sum(scores) / count
highest = max(scores)
lowest = min(scores)
pass_rate = len(passed_names) / count
print(f"人数: {count}")
print(f"平均分: {average:.2f}")
print(f"最高分: {highest}")
print(f"最低分: {lowest}")
print(f"及格率: {pass_rate:.1%}")

print("\n第五部分")
demo = [10, 20, 30]
alias = demo
copied = demo.copy()
print(f"修改前 -> demo={demo}, alias={alias}, copied={copied}")

alias[0] = 999
print(f"修改 alias[0] 后 -> demo={demo}, alias={alias}, copied={copied}")
print(f"alias is demo : {alias is demo}   (同一个对象，改 alias 就是改 demo)")
print(f"copied is demo: {copied is demo}  (copy() 生成新列表，互不影响)")

print("\n第六部分")
student = ("Alice", 85)
print(f"学生记录: {student}")
try:
    student[1] = 90
except TypeError as e:
    print(f"捕获到异常: {e}")
    #原因：元组是不可变序列，不能通过索引重新赋值。

short_names = ["Alice", "Bob", "Carol"]
short_scores = [85, 58]
paired = list(zip(short_names, short_scores))
print(f"长度 3 与 2 配对: {paired}")
print(f"配对个数: {len(paired)}  (zip() 以最短序列为准，多余元素被丢弃)")
```

实验4：⽂件读写、异常处理与模块

```python
#实验四
print("\n第一部分")
import os, pickle, sys, importlib

print(f"当前工作目录: {os.getcwd()}")
DATA_DIR = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(DATA_DIR, exist_ok=True)
print(f"数据目录已就绪: {DATA_DIR}")

print("\n第二部分")
students_path = os.path.join(DATA_DIR, "students.txt")

raw_lines = [
    "name,score",
    "Alice,85",
    "Bob,58",
    "Carol,92",
    "David,76",
    "Eva,60",
    "Frank,89",
    "Grace,abc",
    "Henry,105",
    "Incomplete",
]

with open(students_path, "w", encoding="utf-8") as f:
    f.write("\n".join(raw_lines) + "\n")
print(f"已写入 {students_path}")

with open(students_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

header = lines[0].strip()
data_lines = lines[1:]
print(f"表头: {header}")
print(f"数据行数: {len(data_lines)}")
print(f"前两行解析预览: {data_lines[0].strip().split(',')} | {data_lines[1].strip().split(',')}")

print("\n第三部分")
def parse_student(line):
    parts = line.strip().split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量应为 2，实际为 {len(parts)}")
    name = parts[0].strip()
    score_text = parts[1].strip()
    if name == "":
        raise ValueError("姓名为空")
    try:
        score = int(score_text)
    except ValueError:
        raise ValueError(f"成绩不是整数: '{score_text}'") from None
    if not 0 <= score <= 100:
        raise ValueError(f"成绩超出 0~100 范围: {score}")
    return {"name": name, "score": score}
print("函数 parse_student 已定义，准备解析数据行。")

print("\n第四部分")
valid_records = []
invalid_records = []

for line_no, line in enumerate(data_lines, start=2):
    if line.strip() == "":
        continue
    try:
        record = parse_student(line)
    except ValueError as e:
        invalid_records.append((line_no, line.strip(), str(e)))
        print(f"第 {line_no} 行无效: {line.strip()} -> {e}")
        continue
    valid_records.append(record)

print("\n第五部分")
valid_count = len(valid_records)
invalid_count = len(invalid_records)
average = sum(r["score"] for r in valid_records) / valid_count
pass_rate = sum(1 for r in valid_records if r["score"] >= 60) / valid_count

report_path = os.path.join(DATA_DIR, "report.txt")
with open(report_path, "w", encoding="utf-8") as f:      # w：覆盖，原有内容被清空
    f.write(f"有效记录数: {valid_count}\n")
    f.write(f"无效记录数: {invalid_count}\n")
    f.write(f"平均分: {average:.2f}\n")
    f.write(f"及格率: {pass_rate:.1%}\n")

with open(report_path, "a", encoding="utf-8") as f:      # a：追加，原有内容保留
    f.write("（追加说明：本行由 a 模式写入）\n")

with open(report_path, encoding="utf-8") as f:
    print("---- report.txt 内容 ----")
    print(f.read(), end="")

# 用临时文件单独演示 w 的清空效果，避免弄乱 report.txt
print("\n---- 以下为临时文件内容，用以展示 w 模式的清空效果 ----")
demo_path = os.path.join(DATA_DIR, "mode_demo.txt")
with open(demo_path, "w", encoding="utf-8") as f:
    f.write("第一行（w 模式写入）\n")
with open(demo_path, "a", encoding="utf-8") as f:
    f.write("第二行（a 模式追加）\n")
print("a 模式后:", open(demo_path, encoding="utf-8").read().replace("\n", " / "))
with open(demo_path, "w", encoding="utf-8") as f:
    f.write("\nw 模式重写，前两行都没了")
print("w 模式后:", open(demo_path, encoding="utf-8").read().strip())

print("\n第六部分")
pkl_path = os.path.join(DATA_DIR, "students.pkl")

with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)

with open(pkl_path, "rb") as f:
    loaded_records = pickle.load(f)

print(f"pickle 恢复的数据: {loaded_records}")
print(f"恢复的数据与原数据相等: {loaded_records == valid_records}")

print("\n第七部分")
module_source = '''"""第三部分已经定义的 parse_student 函数，单独保存为 student_utils.py 模块。"""


def parse_student(line):
    """把一行 "name,score" 文本解析为字典；无效记录抛出 ValueError。"""
    parts = line.strip().split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量应为 2，实际为 {len(parts)}")
    name = parts[0].strip()
    score_text = parts[1].strip()
    if name == "":
        raise ValueError("姓名为空")
    try:
        score = int(score_text)
    except ValueError:
        raise ValueError(f"成绩不是整数: '{score_text}'") from None
    if not 0 <= score <= 100:
        raise ValueError(f"成绩超出 0~100 范围: {score}")
    return {"name": name, "score": score}
'''

module_path = os.path.join(os.getcwd(), "student_utils.py")
with open(module_path, "w", encoding="utf-8") as f:
    f.write(module_source)
print(f"模块已写入: {module_path}")

if os.getcwd() not in sys.path:
    sys.path.insert(0, os.getcwd())

import student_utils
importlib.reload(student_utils)

print(f"有效输入 'Alice,85' -> {student_utils.parse_student('Alice,85')}")
try:
    student_utils.parse_student("Henry,105")
except ValueError as e:
    print(f"无效输入 'Henry,105' -> 捕获到异常: {e}")

print("\n第八部分")
missing_path = os.path.join(DATA_DIR, "not_exist.csv")
try:
    with open(missing_path, encoding="utf-8") as f:
        f.read()
except FileNotFoundError as e:
    print(f"捕获到异常: {e}")
    print(f"说明: 文件 {missing_path} 不存在，请检查路径或先创建该文件。")
```

实验5：类、继承与对象复制

```python
#实验五
print("\n第一部分")
class Student:
    """学生：姓名 + 成绩。"""

    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"
print("Student 类已定义完成。")

print("\n第二部分")
students = [Student(r["name"], r["score"]) for r in valid_records]
for s in students:
    status = "及格" if s.is_passed() else "不及格"
    print(f"{s} -> {status}")

print("\n第三部分")
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return (f"GraduateStudent(name={self.name}, score={self.score}, "
                f"research_topic={self.research_topic})")

g = GraduateStudent("Grace", 88, "计算机视觉")
print(g)
print("及格状态:", g.is_passed())
print("是 Student 的实例:", isinstance(g, Student))

print("\n第四部分")
import copy

def make_origin():
    return {"name": "Alice", "scores": [85, 90]}

print("--- 直接赋值 ---")
a = make_origin()
b = a
b["scores"].append(95)
print("原始:", a)
print("别名:", b)
print("同一对象:", a is b, "| 内部列表同一对象:", a["scores"] is b["scores"])

print("\n--- copy.copy 浅拷贝 ---")
a = make_origin()
b = copy.copy(a)
b["scores"].append(95)
print("原始:", a)
print("浅拷贝:", b)
print("同一对象:", a is b, "| 内部列表同一对象:", a["scores"] is b["scores"])

print("\n--- copy.deepcopy 深拷贝 ---")
a = make_origin()
b = copy.deepcopy(a)
b["scores"].append(95)
print("原始:", a)
print("深拷贝:", b)
print("同一对象:", a is b, "| 内部列表同一对象:", a["scores"] is b["scores"])
```
