欧几里得算法

```python
print("\n第一部分")
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m
print("gcd函数已定义")

print("\n第二部分")
for m, n in [(2018, 2019), (1200, 300), (5040, 60)]:
    print(f"gcd({m}, {n}) = {gcd(m, n)}")

print("\n第三部分")
for m, n in [(-2018, 2019), (2018, -2019), (-1200, -300), (-5, 0)]:
    print(f"gcd({m}, {n}) = {gcd(m, n)}")
print("函数不会报错，算出来的绝对值是对的，但符号可能是负的。")

```

#近似欧拉数e

```python
print("\n第一部分")
def euler_limit(n):
    return (1 + 1 / n) ** n
print(f"euler_limit(10) = {euler_limit(10)}")
print(f"euler_limit(1000) = {euler_limit(1000)}")

print("\n第二部分")
def euler_infinite_sum(n):
    total = 0.0
    term = 1.0
    for k in range(n):
        total += term
        term /= (k + 1)
    return total
print(f"euler_infinite_sum(0) = {euler_infinite_sum(0)}")
print(f"euler_infinite_sum(1) = {euler_infinite_sum(1)}")
print(f"euler_infinite_sum(4) = {euler_infinite_sum(4)}")

print("\n第三部分")
import math

def euler_approx(epsilon):
    true_value = math.exp(1)
    n = 0
    approx = euler_infinite_sum(n)
    while abs(approx - true_value) >= epsilon:
        n += 1
        approx = euler_infinite_sum(n)
    return approx
approx = euler_approx(1e-6)
print(f"euler_approx(1e-6) = {approx}")
print(f"math.exp(1)        = {math.exp(1)}")
print(f"误差               = {abs(approx - math.exp(1)):.3e}")

print("\n第四部分")
def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))
print_euler_sum_table(5)
```

测试整数属性

```python
print("\n第一部分")
def is_power_of_2(n):
    while n > 1:
        if n % 2 != 0:
            return False
        n //= 2
    return True

for n in (1, 2, 3, 4, 6, 8, 16, 1024, 1023):
    print(f"{n}: {is_power_of_2(n)}")

print("\n第二部分")
def is_power(b, n):
    if b == 1:
        return n == 1
    while n > 1:
        if n % b != 0:
            return False
        n //= b
    return True

for b, n in [(2, 8), (2, 9), (3, 27), (3, 26), (5, 125), (10, 1000), (1, 5)]:
    print(f"is_power({b}, {n}) = {is_power(b, n)}")
```
