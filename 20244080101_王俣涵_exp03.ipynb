实验1：正则表达式的基础匹配

```python
#实验一
print("\n第一部分")
import re

text = "cat on mat; later: cats and dogs"

print("原文本：", text)
print(r"\d+ 匹配：", re.findall(r"\d+", text))
print(r"\s+ 匹配：", re.findall(r"\s+", text))
print(r"\bcat\b 匹配：", re.findall(r"\bcat\b", text))

print("\n第二部分")
text = "cat dog bird CAT"

print("原文本：", text)
print(r"cat|dog 匹配：", re.findall(r"cat|dog", text))

print(r're.findall(r"a|aa|aaa", "aaaa")：',
      re.findall(r"a|aa|aaa", "aaaa"))
print(r're.findall(r"aaa|aa|a", "aaaa")：',
      re.findall(r"aaa|aa|a", "aaaa"))

print("\n第三部分")
text = "ct cat caat caaat"

patterns = [r"ca*t", r"ca+t", r"ca?t", r"ca{2}t", r"ca{1,2}t"]

print("原文本：", text)
for pattern in patterns:
    print(f"{pattern} 匹配：{re.findall(pattern, text)}")
```

实验2：分组、回溯引⽤与替换

```python
#实验二
print("\n第一部分")
import re

text = "联系：alice@example.com；备注：bob@test.org"

pattern = r"([\w.+-]+)@([\w.-]+\.[A-Za-z]{2,})"
matches = re.findall(pattern, text)

print("原文本：", text)
for username, domain in matches:
    print(f"用户名：{username}，域名：{domain}")

print("\n第二部分")
text = "cat cat, dog cat, bird bird"

pattern = r"\b(\w+)\s+\1\b"
matches = re.findall(pattern, text)

print("原文本：", text)
print("连续重复的单词：", matches)

print("\n第三部分")
text = "2024-05-01\n2024/05/02\n2024.05.03"

pattern = r"\b(\d{4})[-/.](\d{2})[-/.](\d{2})\b"
result = re.sub(pattern, r"\2-\3-\1", text)

print("转换前：")
print(text)
print("转换后：")
print(result)

print("\n第四部分")
text = "联系我：123-456-7890 或 987.654.3210"

pattern = r"(?<!\d)(\d{3})[-. ]?(\d{3})[-. ]?(\d{4})(?!\d)"
result = re.sub(pattern, r"(\1) \2-\3", text)

print("转换前：", text)
print("转换后：", result)
```

实验3：使⽤ BeautifulSoup 解析 HTML

```python
#实验三
from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲：正则表达式</p>
<p class="course">第七讲：结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""

soup = BeautifulSoup(html_doc, "html.parser")

print("\n第一部分")
title = soup.title.get_text()
h1 = soup.find("h1")

print("页面标题：", title)
print("h1 标签名：", h1.name)
print("h1 文本：", h1.get_text())

print("\n第二部分")
print("课程段落：")
for paragraph in soup.find_all("p", class_="course"):
    print(paragraph.get_text())

print("\n链接：")
for link in soup.find_all("a"):
    print("链接文字：", link.get_text())
    print("href：", link.get("href"))

print("\n第三部分")
print("整个文档的可见文本：")
print(soup.get_text(" ", strip=True))

print("\n各段落文本：")
for paragraph in soup.find_all("p"):
    print(paragraph.get_text(strip=True))

print("\n各链接文本：")
for link in soup.find_all("a"):
    print(link.get_text(strip=True))
```
